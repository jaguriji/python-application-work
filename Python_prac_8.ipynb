```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Student": ["David", "Samuel", "Terry", "Evan"],
    "Age": [27, 24, 22, 32],
    "Country": ["UK", "Canada", "China", "USA"],
    "Course": ["Python", "Data Structures", "Machine Learning", "Web Development"],
    "Marks": [85, 72, 89, 76],
}
df = pd.DataFrame(data)
print("Problem 1: DataFrame")
print(df)
```

```python
b = df["Marks"]
print("Problem 2: Marks column (b)")
print(b)
```

```python
c = df[["Country", "Course"]]
print("Problem 3: Country and Course columns (c)")
print(c, "\n")
```

```python
customers = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vikram",
             "Anjali", "Rohan", "Neha", "Karan", "Pooja"],
    "Age": [25, 34, 45, 29, 38, 31, 22, np.nan, 41, 27],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [1500, 2300, 1800, 2700, 3200,
                        2100, np.nan, 1900, 2500, 1700],
}
cust = pd.DataFrame(customers)
 
print("Problem 4: First 5 rows")
print(cust.head(), "\n")
 
print("Summary statistics")
print(cust.describe(), "\n")
 
print("Missing values per column (before handling)")
print(cust.isnull().sum(), "\n")

cust["Age"] = cust["Age"].fillna(cust["Age"].median())
cust["Purchase_Amount"] = cust["Purchase_Amount"].fillna(cust["Purchase_Amount"].median())
 
print("Missing values per column (after handling)")
print(cust.isnull().sum(), "\n")
```

```python
print("Problem 5: Customers with Age > 30")
print(cust[cust["Age"] > 30], "\n")
 
print("Total Purchase_Amount by Gender")
print(cust.groupby("Gender")["Purchase_Amount"].sum(), "\n")
 
print("Customer_ID, Name, Purchase_Amount only")
print(cust[["Customer_ID", "Name", "Purchase_Amount"]], "\n")
```

```python
cust["Discounted_Price"] = cust["Purchase_Amount"] * 0.90
print("Problem 6: With Discounted_Price column")
print(cust, "\n")
 
print("Average Purchase_Amount by Gender")
print(cust.groupby("Gender")["Purchase_Amount"].mean(), "\n")
 
print("Sorted by Purchase_Amount (descending)")
print(cust.sort_values(by="Purchase_Amount", ascending=False))
```
